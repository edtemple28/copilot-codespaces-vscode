```python
# 🧮 System of Equations + Matrix Quiz (Final One-Cell Colab Version)
import random
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# --- Step 1: System Generator ---
def generate_system():
    system_type = random.choice(["one", "none", "infinite"])
    if system_type == "one":
        m1 = random.randint(-3, 3)
        m2 = m1
        while m2 == m1:
            m2 = random.randint(-3, 3)
        b1 = random.randint(-5, 5)
        b2 = random.randint(-5, 5)
        a1, b1c, c1 = -m1, 1, b1
        a2, b2c, c2 = -m2, 1, b2
    elif system_type == "none":
        m = random.randint(-3, 3)
        b1 = random.randint(-5, 5)
        b2 = b1 + random.choice([2, 3, 4])
        a1, b1c, c1 = -m, 1, b1
        a2, b2c, c2 = -m, 1, b2
    else:  # infinite
        m = random.randint(-3, 3)
        b = random.randint(-5, 5)
        a1, b1c, c1 = -m, 1, b
        a2, b2c, c2 = -m * 2, 2, b * 2
    if a1 == 0 and b1c == 0: a1 = 1
    if a2 == 0 and b2c == 0: a2 = 1
    return a1, b1c, c1, a2, b2c, c2, system_type

# --- Step 2: Plot Graphs (extra spacing after question) ---
def plot_graph():
    x = np.linspace(-10, 10, 200)
    fig, axs = plt.subplots(1, 3, figsize=(12, 3))

    axs[0].plot(x, 2 * x + 1, label="Eq1")
    axs[0].plot(x, -x + 3, label="Eq2")
    axs[0].set_title("Graph 1: Intersecting")
    axs[0].legend()

    axs[1].plot(x, x + 1, label="Eq1")
    axs[1].plot(x, x + 4, label="Eq2")
    axs[1].set_title("Graph 2: Parallel")
    axs[1].legend()

    axs[2].plot(x, 0.5 * x + 2, label="Eq1 & Eq2")
    axs[2].set_title("Graph 3: Same Line")
    axs[2].legend()

    plt.tight_layout()
    plt.show()

    # Added extra spacing for clarity
    print("\n2️⃣ Which graph matches the system?\n\n\n\n")

# --- Step 3: Matrix Question ---
def show_matrix(a1, b1, c1, a2, b2, c2, system_type):
    print("\n🧮 Augmented Matrix:")
    print(f"[ {a1:>3}  {b1:>3} | {c1:>3} ]")
    print(f"[ {a2:>3}  {b2:>3} | {c2:>3} ]")
    print("\nAfter eliminating x from the 2nd row, what happens?")
    options = [
        ("Becomes [0 0 | 0]  (all zeros)", "zero"),
        ("Becomes [0 0 | nonzero]  (inconsistent)", "inconsistent"),
        ("Becomes normal numbers (still solvable)", "normal")
    ]
    if system_type == "infinite":
        correct = "zero"
    elif system_type == "none":
        correct = "inconsistent"
    else:
        correct = "normal"
    widget = widgets.RadioButtons(options=options, description="")
    display(widget)
    return widget, correct

# --- Step 4: One Question Interactive ---
def one_question(q_num, score):
    clear_output()
    print(f"\n=============================")
    print(f"🧠 QUESTION {q_num}")
    print(f"=============================")
    a1, b1, c1, a2, b2, c2, stype = generate_system()
    print(f"System:\n {a1}x + {b1}y = {c1}\n {a2}x + {b2}y = {c2}")

    if stype == "infinite":
        correct_type, correct_graph = "1", "3"
    elif stype == "none":
        correct_type, correct_graph = "3", "2"
    else:
        correct_type, correct_graph = "2", "1"

    print("\n1️⃣ What kind of solution does this system have?")
    type_widget = widgets.RadioButtons(
        options=[('Infinite Solutions', '1'), ('One Solution', '2'), ('No Solution', '3')],
        description=''
    )
    display(type_widget)

    plot_graph()
    graph_widget = widgets.RadioButtons(
        options=[('Graph 1', '1'), ('Graph 2', '2'), ('Graph 3', '3')],
        description=''
    )
    display(graph_widget)

    matrix_widget, matrix_correct = show_matrix(a1, b1, c1, a2, b2, c2, stype)

    submit_btn = widgets.Button(description="Submit Answers ✅", button_style='success')
    next_btn = widgets.Button(description="Next Question ➡️", button_style='info')
    output = widgets.Output()
    display(submit_btn, output)

    def check_answers(b):
        with output:
            clear_output()
            if (type_widget.value == correct_type and
                graph_widget.value == correct_graph and
                matrix_widget.value == matrix_correct):
                print("🌟 You are now an expert! 🌟")
                score[0] += 1
            else:
                print("❌ Not quite. Review your steps and try again.")
            submit_btn.disabled = True
            display(next_btn)

    def next_question(b):
        if q_num < 3:
            one_question(q_num + 1, score)
        else:
            final_results(score)

    submit_btn.on_click(check_answers)
    next_btn.on_click(next_question)

# --- Step 5: Final Results ---
def final_results(score):
    clear_output()
    print("\n=============================")
    print(f"✅ You got {score[0]} out of 3 correct.")
    if score[0] == 3:
        print("🎉 You pass! Show your teacher your results.")
    else:
        print("❌ Try again — your teacher still believes in you.")
    print("=============================")

    restart_btn = widgets.Button(description="Restart 🔁", button_style='warning')
    display(restart_btn)
    restart_btn.on_click(lambda b: quiz())

# --- Step 6: Run Quiz ---
def quiz():
    score = [0]
    one_question(1, score)

quiz()
```
